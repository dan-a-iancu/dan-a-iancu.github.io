# 4. Comparisons, Boolean logic, and `if` statements


Comparisons produce `True` or `False`.


````python
demand = 75
print(demand == 75)
print(demand != 50)
print(demand > 50)
print(demand <= 30)
````

> **REMEMBER:** `=` assigns a value. `==` compares two values. Mixing these up is a very common error.


You can combine conditions with `and`, `or`, and `not`. Add parentheses for each logical condition, to help with readability.


````python
capacity = 100
price = 40
print( (demand <= capacity) and (price > 0) )
print( (demand > capacity) or (price < 0) )
print(not (demand > capacity))
````

Use `in` to test membership in a collection.


````python
products = ["Basic", "Premium", "Deluxe"]
print("Premium" in products)
print("Economy" not in products)
````

Conditional logic uses `if`, optionally followed by `elif` and `else`. Indentation is part of Python syntax.


````python
profit = 125

if profit > 0:
    print("Profitable")
elif profit == 0:
    print("Break-even")
else:
    print("Unprofitable")
````

> **REMEMBER:** Python requires a colon `:` after `if`, `elif`, `else`, `for`, and `def`. The indented block is the code controlled by that statement.


You will also occasionally see a compact **conditional expression**:


````python
status = "profitable" if profit > 0 else "not profitable"
print(status)
````

> **RECOGNIZE / LOOK UP AS NEEDED:** Read `A if condition else B` as: use A when the condition is true; otherwise use B.

# 5. `for` loops, `range`, and `enumerate`


A `for` loop repeats a block of code once for each item in a collection.


````python
products = ["Basic", "Premium", "Deluxe"]

for p in products:
    print(p)
````

> **REMEMBER:** Read `for p in products:` as “for each value in `products`, temporarily call that value `p` and execute the indented block.”


`range(n)` generates the integers `0, 1, ..., n-1`. The endpoint is excluded.


````python
for i in range(5):
    print(i)
````

````python
for i in range(2, 6):
    print(i)
````

`enumerate(...)` is useful when you need both the position and the value.


````python
for i, p in enumerate(products):
    print(i, p)
````

> **REMEMBER:** Nested `for` loops appear in many course notebooks. Remember that for each value of the outer loop, the inner loop completes all its iterations.


````python
days = ["Mon", "Tue"]
products = ["Basic", "Premium"]

for day in days:
    for p in products:
        print(day, p)
````

# 6. Lists: ordered collections


A list stores an ordered collection of values.


````python
demands = [80, 95, 110, 70]
print(demands)
print(len(demands))
````

> **REMEMBER:** Python indexing starts at 0. `x[0]` is the first element, `x[-1]` is the last.


````python
print(demands[0])
print(demands[2])
print(demands[-1])
````

A slice retrieves several consecutive elements. As with `range`, the ending position is excluded.


````python
print(demands[1:3])
print(demands[:2])
print(demands[2:])
````

In Python, a list can contain elements that are different data types.

````python
crazy_list = ["a string", 4.75, ['Federer', 'Nadal', 'Djokovic'], 5, True]
print(crazy_list)
````

You can check the **length** of the list---how many elements are in the list---using the `len` function.

````python
print(f"List has {len(crazy_list)} elements.")
````

> **REMEMBER:** Use `len` to check the length of a list.

Lists are mutable: you can change them.


````python
inventory = [10, 15, 12]
inventory[1] = 20
inventory.append(8)
print(inventory)
last = inventory.pop()
print(last, inventory)
````

> **REMEMBER:** Use `.copy()` when you need a separate copy of a list rather than a second name for the same list.


````python
a = [1, 2, 3]
b = a.copy()
b[0] = 99
print(a)
print(b)
````

Useful built-ins for numeric lists include `sum`, `min`, and `max`.


````python
values = [12, 8, 17, 10]
print(sum(values))
print(min(values))
print(max(values))
````

For Python lists, `+` means concatenation and `*` means repetition:

````python
a = [1, 2]
b = [3, 4]

print(a + b)
# [1, 2, 3, 4]

print(3 * a)
# [1, 2, 1, 2, 1, 2]
````

## List comprehensions


A list comprehension creates a list by applying an expression to each element of another collection.


````python
demands = [80, 95, 110, 70]
squared_demands = [d**2 for d in demands]
print(squared_demands)
````

> **REMEMBER:** Read `[expression for x in items]` as “create a list containing `expression`, once for each `x` in `items`."


You can filter within a comprehension:


````python
high_demands = [d for d in demands if d >= 90]
print(high_demands)
````

A closely related pattern is a generator expression inside `sum(...)`. **You will see this frequently in OIT 248 code.**


````python
prices = [10, 15, 20]
quantities = [3, 4, 2]
total_revenue = sum(prices[i] * quantities[i] for i in range(len(prices)))
print(total_revenue)
````

> **REMEMBER:** Read `sum(expression for x in items)` as the sum of all the values in the collection `items` (`x` is just a local variable that iterates through those items)

# 8. Dictionaries: values indexed by meaningful keys


A dictionary stores **key-value pairs**. Dictionaries are extremely useful when your data are naturally indexed by names, products, locations, time periods, etc.


````python
price = {"Basic": 20, "Premium": 35, "Deluxe": 50}
print(price["Premium"])
````

> **REMEMBER:** Lists use integer positions; dictionaries use keys. For example, `price["Premium"]` retrieves the value associated with the key `"Premium"`.


````python
price["Economy"] = 15
print(price)
````

You can use the `.items()` method to iterate through all the key-value pairs in a dictionary.

````python
for product, p in price.items():
    print(product, p)
````

To obtain the keys or values of a dictionary, use the `.keys()` and `.values()` methods. These return special objects that you can convert to regular Python lists with the `list()` function.

````python
# Get all keys
product_types = price.keys()
print(product_types)
print(list(product_types))

# Get all values
product_prices = price.values()
print(product_prices)
print(list(product_prices))
````

 > Use `.keys()` to get all dictionary keys and `.values()` to get all dictionary values. If needed, wrap the result in `list(...)` to convert it to a regular list.<br>
> Use `items()` to iterate through all the key-value pairs.

Dictionary comprehensions are the dictionary analogue of list comprehensions.


````python
products = ["Basic", "Premium", "Deluxe"]
unit_cost = {p: 0.6 * price[p] for p in products}
print(unit_cost)
````

# 9. Functions


Functions package reusable calculations under a name.


````python
def contribution_margin(price, unit_cost, quantity):
    margin = (price - unit_cost) * quantity
    return margin

print(contribution_margin(32, 19, 250))
````

> **REMEMBER:** A function definition starts with `def`, lists inputs in parentheses, uses an indented body, and sends a result back with `return`.


Functions can have **default arguments**:


````python
def revenue(price, quantity=1):
    return price * quantity

print(revenue(20))
print(revenue(20, 5))
print(revenue(price=20, quantity=5))
````

> **RECOGNIZE / LOOK UP AS NEEDED:** Arguments can be supplied by position or by name. Default arguments let the caller omit an input when the default is appropriate.


Functions can return multiple values, which are commonly unpacked immediately:


````python
def summary(values):
    return min(values), max(values), sum(values) / len(values)

low, high, avg = summary([4, 7, 1, 10])
print(low, high, avg)
````

`None` means “no value.” It is often used as a default when an argument is optional.


````python
def greet(name=None):
    if name is None:
        return "Hello"
    return f"Hello, {name}"

print(greet())
print(greet("Ada"))
````

# 15. NumPy: arrays and numerical calculations

NumPy is useful for numerical arrays and simulation. A NumPy array resembles a list but supports efficient element-by-element numerical operations.


````python
x = np.array([10, 20, 30, 40])
print(x)
print(x[0])
print(x[1:3])
````

Arithmetic with an array is vectorized:


````python
print(x + 5)
print(2 * x)
print(x / 10)
````

> **REMEMBER:** For ordinary Python lists, `2 * [1, 2]` repeats the list. For NumPy arrays, `2 * np.array([1, 2])` multiplies the values. Know which kind of object you are working with.


Calculations can also combine arrays of the same size, operating on corresponding elements.


````python
prices = np.array([20, 35, 50])
quantities = np.array([100, 60, 25])
revenues = prices * quantities

print(revenues)
print(np.sum(revenues))
````

> **RECOGNIZE / LOOK UP AS NEEDED:** When two arrays have compatible sizes, arithmetic such as `x * y` is performed element by element.


## Array constructors

Some common ways to create arrays are:


````python
print(np.zeros(4))              # array with 4 zeros
print(np.ones(3))               # array with 3 ones
print(np.arange(0, 3, 0.25))    # array with values from 0 to 3 (excluding), in increments of 0.25
print(np.linspace(0, 1, 6))     # array with 6 evenly-spaced values from 0 to 1 (including)
````

## Minimum, maximum, and their positions

NumPy distinguishes between the smallest or largest **value** and the **position** where that value occurs.


````python
values = np.array([3, 7, 2, 10, 6])

print(f"Minimum value: {np.min(values)}")
print(f"Position of minimum: {np.argmin(values)}")
print(f"Maximum value: {np.max(values)}")
print(f"Position of maximum: {np.argmax(values)}")
````

> **REMEMBER:** `np.min(x)` and `np.max(x)` return the smallest and largest **values**. `np.argmin(x)` and `np.argmax(x)` return their zero-based **positions**. If the extreme value occurs more than once, `argmin` or `argmax` returns the first such position.


## Logical comparisons, counts, and proportions

A comparison involving an entire NumPy array is applied separately to every element. The result is a Boolean array containing `True` and `False`.


````python
demand = np.array([82, 101, 94, 130, 75, 118])
above_100 = demand > 100

print(above_100)
print(f"Number above 100: {np.sum(above_100)}")
print(f"Proportion above 100: {np.mean(above_100):.3f}")
````

> **REMEMBER:** A comparison such as `x > 100` produces one `True`/`False` value for every element of `x`. Because NumPy treats `True` as 1 and `False` as 0, `np.sum(condition)` counts how many times the condition holds and `np.mean(condition)` calculates the proportion of times it holds.


### Filtering and combining conditions

A Boolean array can be used to select only the array elements satisfying a condition. Conditions can also be combined.


````python
in_target_range = (demand >= 80) & (demand <= 110)
outside_target_range = (demand < 80) | (demand > 110)
not_above_100 = ~(demand > 100)

print(demand[in_target_range])
print(outside_target_range)
print(not_above_100)
````

> **RECOGNIZE / LOOK UP AS NEEDED:** For NumPy arrays, combine conditions using `&` (and), `|` (or), and `~` (not), rather than Python's `and`, `or`, and `not`. Put each individual comparison in parentheses. The expression `x[condition]` retrieves the elements for which the condition is `True`.


### Other element-by-element choices

NumPy also provides functions for making an element-by-element minimum, maximum, or conditional choice.


````python
capacity = 100
sales = np.minimum(demand, capacity)
shortfall = np.maximum(demand - capacity, 0)
status = np.where(demand > capacity, "Stockout", "Enough capacity")

print(sales)
print(shortfall)
print(status)
````

> **RECOGNIZE / LOOK UP AS NEEDED:** `np.minimum(x, y)` and `np.maximum(x, y)` compare corresponding values. `np.where(condition, value_if_true, value_if_false)` selects between two alternatives for each array element.


## Means, standard deviations, covariance, and correlation

NumPy can also calculate common statistical summaries.


````python
price = np.array([10, 12, 14, 16, 18])
demand_for_stats = np.array([110, 102, 95, 87, 80])

mean_demand = np.mean(demand_for_stats)
population_std = np.std(demand_for_stats)
sample_std = np.std(demand_for_stats, ddof=1)
covariance = np.cov(price, demand_for_stats)[0, 1]
correlation = np.corrcoef(price, demand_for_stats)[0, 1]

print(f"Mean demand: {mean_demand:.2f}")
print(f"Population standard deviation: {population_std:.2f}")
print(f"Sample standard deviation: {sample_std:.2f}")
print(f"Covariance of price and demand: {covariance:.2f}")
print(f"Correlation of price and demand: {correlation:.3f}")
````

> **RECOGNIZE / LOOK UP AS NEEDED:** Use `np.mean(x)` for the mean and `np.std(x)` for the population standard deviation. Use `np.std(x, ddof=1)` for the sample standard deviation. `np.cov(x, y)` uses sample covariance by default. `np.cov(x, y)` and `np.corrcoef(x, y)` return 2-by-2 matrices; the entry `[0, 1]` is the covariance or correlation between `x` and `y`.


Other NumPy functions that appear occasionally in OIT 248 include `np.floor`, `np.round`, `np.sort`, `np.unique`, and `np.meshgrid`. Look them up when needed rather than memorizing all of them.

# 16. Random numbers and simulation

Simulation requires random samples. In OIT 248 we will use NumPy's modern random-number generator.


````python
rng = np.random.default_rng(seed=248)
````

> **REMEMBER:** Create one generator with `rng = np.random.default_rng(seed=...)`, then call methods such as `rng.normal(...)`. A fixed seed makes the same random experiment reproducible.


Some distributions you will encounter:


````python
normal_samples = rng.normal(loc=100, scale=15, size=5)    # 5 values drawn from a normal distribution with mean 100 and standard deviation 15
integer_samples = rng.integers(low=1, high=7, size=5)     # 5 values drawn from a discrete uniform distribution with potential values 1,2,...,6
binary_samples = rng.binomial(n=1, p=0.3, size=10)        # 10 values drawn from a binomial distribution with n=1 trials and probability of success p=0.3
choices = rng.choice(["Low", "Medium", "High"], size=5, p=[0.2, 0.5, 0.3])    # 5 values drawn from a custom distribution

print(normal_samples)
print(integer_samples)
print(binary_samples)
print(choices)
````

> **RECOGNIZE / LOOK UP AS NEEDED:** The exact arguments vary by distribution. In the course you may also see `rng.triangular(...)` and `rng.multivariate_normal(...)`; look up those signatures when needed.


A basic simulation often has the pattern **generate → calculate outcome → summarize**.


````python
rng = np.random.default_rng(seed=248)
demand = rng.normal(loc=100, scale=20, size=10_000)
capacity = 110
sales = np.minimum(demand, capacity)

print(f"Mean demand: {np.mean(demand):.2f}")
print(f"Mean sales:  {np.mean(sales):.2f}")
print(f"Stockout probability: {np.mean(demand > capacity):.3f}")
````
